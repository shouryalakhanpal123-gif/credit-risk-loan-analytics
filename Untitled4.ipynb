```python
import pandas as pd
import numpy as np

transactions["transaction_date"] = pd.to_datetime(
    transactions["transaction_date"],
    errors="coerce"
)

loan_applications["application_date"] = pd.to_datetime(
    loan_applications["application_date"],
    errors="coerce"
)

print(transactions["transaction_date"].min())
print(transactions["transaction_date"].max())

print(loan_applications["application_date"].min())
print(loan_applications["application_date"].max())
```

```python
print(transactions["transaction_type"].value_counts(dropna=False))
print(transactions["category"].value_counts(dropna=False))
```

```python
# Transaction data ki copy
txn = transactions.copy()

# Dates valid honi chahiye
txn = txn.dropna(
    subset=["customer_id", "transaction_date"]
)

# Amount numeric hona chahiye
txn["amount"] = pd.to_numeric(
    txn["amount"],
    errors="coerce"
)

txn["balance_after"] = pd.to_numeric(
    txn["balance_after"],
    errors="coerce"
)

# Transaction type normalize karo
txn["transaction_type_clean"] = (
    txn["transaction_type"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# Credit aur debit amounts
txn["inflow"] = np.where(
    txn["transaction_type_clean"].eq("credit"),
    txn["amount"],
    0
)

txn["outflow"] = np.where(
    txn["transaction_type_clean"].eq("debit"),
    txn["amount"],
    0
)

# EMI outflow
txn["category_clean"] = (
    txn["category"]
    .astype("string")
    .str.strip()
    .str.lower()
)

txn["emi_outflow"] = np.where(
    txn["transaction_type_clean"].eq("debit")
    & txn["category_clean"].str.contains(
        "emi", na=False
    ),
    txn["amount"],
    0
)
```

```python
applications = loan_applications[
    ["application_id", "customer_id", "application_date"]
].copy()

applications = applications.dropna(
    subset=["application_id", "customer_id", "application_date"]
)

applications["window_start"] = (
    applications["application_date"] - pd.DateOffset(months=6)
)

# Customer aur application date ke basis par transaction window
txn_joined = applications.merge(
    txn,
    on="customer_id",
    how="left"
)

# Sirf application date se pehle ke 6 months
txn_joined = txn_joined[
    (txn_joined["transaction_date"] >= txn_joined["window_start"])
    & (txn_joined["transaction_date"] < txn_joined["application_date"])
]

transaction_features = (
    txn_joined
    .groupby("application_id")
    .agg(
        transaction_count_6m=("transaction_id", "nunique"),
        total_inflow_6m=("inflow", "sum"),
        total_outflow_6m=("outflow", "sum"),
        net_cash_flow_6m=("amount", "size"),
        average_balance_6m=("balance_after", "mean"),
        emi_outflow_6m=("emi_outflow", "sum")
    )
    .reset_index()
)

# Net cash flow ko sahi formula se calculate karo
transaction_features["net_cash_flow_6m"] = (
    transaction_features["total_inflow_6m"]
    - transaction_features["total_outflow_6m"]
)

# Inflow-to-outflow ratio
transaction_features["inflow_to_outflow_ratio_6m"] = np.where(
    transaction_features["total_outflow_6m"] > 0,
    transaction_features["total_inflow_6m"]
    / transaction_features["total_outflow_6m"],
    np.nan
)

print(transaction_features.head())
print(transaction_features.shape)
```

```python
ml_df = loan_applications.merge(
    customers,
    on="customer_id",
    how="left",
    suffixes=("", "_customer")
)

# Credit history mein customer ke multiple records hain ya nahi
print(
    "Credit history records per customer:",
    credit_history.groupby("customer_id").size().max()
)
```

```python
ml_df = ml_df.merge(
    credit_history,
    on="customer_id",
    how="left",
    suffixes=("", "_credit")
)
```

```python
ml_df = ml_df.merge(
    transaction_features,
    on="application_id",
    how="left"
)

print("Final dataset shape:", ml_df.shape)
print(ml_df.head())
```

```python
leakage_columns = [
    "approval_status",
    "approved_amount",
    "decision_date",
    "disbursed_flag"
]

ml_df = ml_df.drop(
    columns=leakage_columns,
    errors="ignore"
)
```

```python
ml_df = ml_df[
    ml_df["loan_default"].notna()
].copy()

ml_df["loan_default"] = ml_df[
    "loan_default"
].astype(int)

print(ml_df["loan_default"].value_counts())
```

```python
y = ml_df["loan_default"]

# Identifiers aur target ko features se remove karo
columns_to_drop = [
    "loan_default",
    "customer_id",
    "application_id",
    "credit_id",
    "transaction_id",
    "first_name",
    "last_name",
    "email",
    "phone"
]

X = ml_df.drop(
    columns=columns_to_drop,
    errors="ignore"
)
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts(normalize=True))
```

```python
numeric_features = X.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool", "string"]
).columns.tolist()

print("Numeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="constant", fill_value="Unknown")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),

    "XGBoost": XGBClassifier(
        n_estimators=300,
        max_depth=4,
        learning_rate=0.05,
        eval_metric="logloss",
        random_state=42
    )
}
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = [
    "accuracy",
    "precision",
    "recall",
    "f1",
    "roc_auc"
]

cv_results = []

for model_name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": model_name,
        "Accuracy": scores["test_accuracy"].mean(),
        "Precision": scores["test_precision"].mean(),
        "Recall": scores["test_recall"].mean(),
        "F1": scores["test_f1"].mean(),
        "ROC_AUC": scores["test_roc_auc"].mean()
    })

cv_results_df = pd.DataFrame(cv_results)

cv_results_df.sort_values(
    "ROC_AUC",
    ascending=False
)
```

```python
best_model_name = "Random Forest"

best_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", models[best_model_name])
])

best_model.fit(X_train, y_train)

print("Best model trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

y_pred = best_model.predict(X_test)

y_probability = best_model.predict_proba(X_test)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, zero_division=0))
print("Recall:", recall_score(y_test, y_pred, zero_division=0))
print("F1 Score:", f1_score(y_test, y_pred, zero_division=0))
print("ROC-AUC:", roc_auc_score(y_test, y_probability))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```
